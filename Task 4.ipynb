In this task, you will work with the dataset provided in the file **`data_banknote_authentication.csv`**

Your objective is to build and evaluate classification models to predict the authenticity of banknotes.

You should:

- Load and explore the dataset.
- Build two classification models:
  - **Decision Tree**
  - **Random Forest**
- Use **GridSearchCV** to optimize the hyperparameters of each model.
- Evaluate the performance of both models using:
  - **Confusion Matrix**
  - **Classification Report**

Finally, **compare the performance of the two models** and discuss which model performs better for this dataset.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
df = pd.read_csv("data_banknote_authentication.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df['Class'].unique()
```

```python
# Features and target
X = df.drop("Class", axis=1)
y = df["Class"]

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.3, random_state=101)
```

```python
# Decision Tree model
model_dt = DecisionTreeClassifier()

# Hyperparameter grid
dt_params = {
    "max_depth": [None, 3, 5, 10],
    "min_samples_split": [2, 5, 10],
    "criterion": ["gini", "entropy"]
}

# Grid search
dt_grid = GridSearchCV(
    model_dt,
    dt_params,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

dt_grid.fit(X_train, y_train)

print("Best Decision Tree Parameters:")
print(dt_grid.best_params_)
```

```python
# Predictions for Decision Tree (tuned model)
dt_pred = dt_grid.predict(X_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, dt_pred))


print("Decision Tree Confusion Matrix:")
print(confusion_matrix(y_test, dt_pred))

print("\nDecision Tree Classification Report:")
print(classification_report(y_test, dt_pred))
```

```python
model_rf = RandomForestClassifier()

rf_params = {
    "n_estimators": [50, 100, 200],
    "max_depth": [None, 5, 10],
    "min_samples_split": [2, 5],
    "max_features": ["sqrt", "log2", None]  # Add this
}

rf_grid = GridSearchCV(
    model_rf,
    rf_params,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

print("Best Random Forest Parameters:")
print(rf_grid.best_params_)
```

```python
preds = rf_grid.predict(X_test)

print("Random Forest Accuracy:", accuracy_score(y_test, preds))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, preds))

print("\nClassification Report:")
print(classification_report(y_test, preds))
```

Decision Tree achieved an accuracy of 99.76%, while the Random Forest achieved 99.03%. 
The Decision Tree uses no restriction on tree depth (max_depth=None), which leads to overfitting so the Random Forest can be a robust model as it reduces the overfitting because it combines multiple decision trees
